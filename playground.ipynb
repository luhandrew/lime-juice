```python
from skimage.segmentation import slic
from skimage.color import label2rgb
from skimage.data import coffee 
import matplotlib.pyplot as plt 

img = coffee()

# documentation:
# - compactness is a measure of how "pixelated" the segments are
# - n_segments gives the number of such pixels

segments = slic(img, compactness=100, n_segments=400)
segments_img = label2rgb(segments, img, kind="avg", bg_label=0)

# display the images
fig, ax = plt.subplots(nrows=2, sharex=True, sharey=True, figsize=(6, 8))
ax[0].imshow(img)
ax[1].imshow(segments_img)
for a in ax:
  a.axis("off")
plt.tight_layout()
```
