```python
import torch 
from torchvision import models
from lime import LimeImageExplainer

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[debug] using device: {device}")

weights = models.ResNet18_Weights.DEFAULT
model = models.resnet18(weights=weights)
print(f"[debug] loaded pre-trained model!")

explainer = LimeImageExplainer(model, device)
print(f"[debug] loaded explainer!")
```

## **Sample examples**

```python
explainer.explain("./images/goose.jfif").visualize()
```

```python
explainer.explain("./images/lemons.jfif").visualize()
```

```python
explainer.explain("./images/red_panda.jfif").visualize()
```

## **Trying different seeds**
The seed shouldn't play that huge of a role in the performance of our algorithm. It should affect it only by a little bit.

```python
explainer.explain("./images/wolf.jfif", random_state=0).visualize()
```

```python
explainer.explain("./images/wolf.jfif", random_state=42).visualize()
```

```python
explainer.explain("./images/wolf.jfif", random_state=61).visualize()
```

## **Playing with the number of samples**
The number of samples plays a huge role in the speed of our algorithm, but also in the model's fidelity. This is a trade-off. The lower this is, the faster our algorithm is, but the worse it performs. The bigger it is, the slower our algorithm is, but it performs way better.

```python
explainer.explain("./images/corn.jfif", num_samples=10).visualize()
```

```python
explainer.explain("./images/corn.jfif", num_samples=100).visualize()
```

```python
explainer.explain("./images/corn.jfif", num_samples=500).visualize()
```

```python
explainer.explain("./images/corn.jfif", num_samples=1000).visualize()
```

```python
explainer.explain("./images/corn.jfif", num_samples=5000).visualize()
```
